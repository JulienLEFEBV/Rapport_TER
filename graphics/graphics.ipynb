```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = lambda x : np.cos(x) + np.exp(-x**2) * 63 * x**3
# -sin(x) + 189x^(2) ℯ^(-x^(2)) - 126x^(4) ℯ^(-x^(2))
fdiff = lambda x : -np.sin(x) + 189*x**2 + np.exp(-x**2) - 126*x**4 * np.exp(-x**2)

h = 0.00000000000001
fdiff_aprox = lambda x : (f(x+h) -f(x))/h

X = np.arange(-1,1,0.001)

plt.plot(X, f(X))
plt.plot(X, fdiff(X))
plt.plot(X, fdiff_aprox(X))
plt.grid()

plt.show()

def MSE(f_ref, f_aproc, X = np.arange(-1,1, 0.00001)):
    return np.mean((f_ref(X) - f_aproc(X))**2)

fdiff_aprox = lambda h : lambda x :  (f(x+h) -f(x))/h
# H = np.arange(1e-16,1e-5, 1e-6)
H = np.linspace(1e-16, 1, 10_000)
mse = np.array([MSE(fdiff, fdiff_aprox(h)) for h in H])
plt.plot(H, mse)
plt.show()
```
